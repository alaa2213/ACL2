# 1. Data Cleaning and Auditing

The Formula 1 dataset consists of 14 relational CSV tables. Before constructing the
modeling dataset, the raw data is explored and audited to identify missing values,
incorrect data types, duplicate records, and relational-integrity problems.

The cleaning process includes:
- exploratory data analysis before cleaning;
- detection and cleaning of `\N` sentinel values;
- semantic typing of dates and durations;
- primary-key and foreign-key validation;
- investigation and resolution of duplicate `(raceId, driverId)` records;
- and exploratory analysis after cleaning.

The original DataFrames are preserved unchanged. All cleaning operations are
performed on separate copies.

## 1.2 Sentinel and Missing-Value Analysis

The Formula 1 CSV files use the literal string `\N` as a sentinel representing
missing or unavailable information.

Because `\N` is a text placeholder rather than a proper missing value, its frequency
is measured before cleaning. Both the number and percentage of sentinel values are
calculated for each affected column and visualized before any replacement is made.

```python
sentinel_summary = []

for table_name, df in raw_data.items():
    for column in df.columns:

        count = (
            df[column]
            .astype(str)
            .str.strip()
            .eq(r"\N")
            .sum()
        )

        if count > 0:
            sentinel_summary.append({
                "Table": table_name,
                "Column": column,
                "Count": count,
                "Percentage": count / len(df) * 100
            })

sentinel_summary = pd.DataFrame(sentinel_summary)

sentinel_summary = sentinel_summary.sort_values(
    "Percentage",
    ascending=False
).reset_index(drop=True)

display(sentinel_summary)
```

### Total Missingness Before Cleaning

The previous analysis counts only the literal `\N` sentinel values. Some columns also
contain values that Pandas already recognizes as missing (`NaN`).

Therefore, total missingness is calculated using both existing null values and `\N`
sentinels to obtain the complete missing-data picture before cleaning.

```python
missing_before = []

for table_name, df in raw_data.items():
    for column in df.columns:

        null_count = df[column].isnull().sum()

        sentinel_count = (
            df[column]
            .astype(str)
            .str.strip()
            .eq(r"\N")
            .sum()
        )

        total_missing = null_count + sentinel_count

        if total_missing > 0:
            missing_before.append({
                "Table": table_name,
                "Column": column,
                "Missing_Count": total_missing,
                "Missing_Percentage": total_missing / len(df) * 100
            })

missing_before = pd.DataFrame(missing_before)

missing_before = missing_before.sort_values(
    "Missing_Percentage",
    ascending=False
).reset_index(drop=True)

display(missing_before.head(30))
```

```python
missing_before_plot = (
    missing_before
    .sort_values(
        "Missing_Percentage",
        ascending=True
    )
    .copy()
)

missing_before_plot["Feature"] = (
    missing_before_plot["Table"]
    + "."
    + missing_before_plot["Column"]
)

plt.figure(
    figsize=(
        12,
        max(6, len(missing_before_plot) * 0.35)
    )
)

ax = sns.barplot(
    data=missing_before_plot,
    x="Missing_Percentage",
    y="Feature"
)

plt.title("Missingness Across All Affected Columns Before Cleaning")
plt.xlabel("Missing Values (%)")
plt.ylabel("")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=3
    )

plt.tight_layout()
plt.show()
```

## 1.4 Cleaning Sentinel Values

All `\N` sentinel values are converted to `NaN`, Pandas' standard representation of
missing data.

Since the final objective is to predict whether a driver finishes on the podium, not
every column in the 14 source tables will be used as a predictive feature. Therefore,
missing values are not dropped or imputed at this stage. Instead, they are represented
correctly as `NaN`, and their treatment will be determined later for the features
selected for the final modeling table.

```python
for name in clean_data:
    clean_data[name] = clean_data[name].replace(
        r"^\s*\\N\s*$",
        np.nan,
        regex=True
    )

print("Sentinel replacement complete.")
```

```python
remaining_sentinels = 0

for name, df in clean_data.items():
    for column in df.columns:

        remaining_sentinels += (
            df[column]
            .astype(str)
            .str.strip()
            .eq(r"\N")
            .sum()
        )

print("Remaining \\N sentinel values:", remaining_sentinels)
```

## 1.5 Semantic Data Types

Several fields are stored as strings even though they represent dates or durations.

Race-event dates and driver dates of birth are converted to datetime values. Genuine
duration fields, such as qualifying lap times, fastest-lap times, lap times, and pit-stop
durations, are converted to numeric seconds.

The original duration columns are retained while new `_seconds` columns are created,
preserving the original representation for auditing.

`results.time` and `sprint_results.time` are not converted using the generic duration
function because these columns mix full elapsed race times with time gaps such as
`+5.478`, which do not have the same semantic meaning.

```python
race_date_columns = [
    "date",
    "fp1_date",
    "fp2_date",
    "fp3_date",
    "quali_date",
    "sprint_date"
]

for column in race_date_columns:
    clean_data["races"][column] = pd.to_datetime(
        clean_data["races"][column],
        errors="coerce"
    )

clean_data["drivers"]["dob"] = pd.to_datetime(
    clean_data["drivers"]["dob"],
    errors="coerce"
)

print("Date conversion complete.")
```

```python
numeric_columns = {
    "results": [
        "position",
        "milliseconds",
        "fastestLap",
        "rank",
        "fastestLapSpeed"
    ],
    "sprint_results": [
        "position",
        "milliseconds",
        "fastestLap"
    ]
}

for table_name, columns in numeric_columns.items():
    for column in columns:
        clean_data[table_name][column] = pd.to_numeric(
            clean_data[table_name][column],
            errors="coerce"
        )

print("Numeric semantic typing complete.")
```

```python
def time_to_seconds(value):

    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value == "":
        return np.nan

    parts = value.split(":")

    try:
        if len(parts) == 1:
            return float(parts[0])

        elif len(parts) == 2:
            minutes = float(parts[0])
            seconds = float(parts[1])

            return minutes * 60 + seconds

        elif len(parts) == 3:
            hours = float(parts[0])
            minutes = float(parts[1])
            seconds = float(parts[2])

            return (
                hours * 3600
                + minutes * 60
                + seconds
            )

    except ValueError:
        return np.nan

    return np.nan
```

```python
duration_columns = {
    "results": [
        "fastestLapTime"
    ],

    "qualifying": [
        "q1",
        "q2",
        "q3"
    ],

    "lap_times": [
        "time"
    ],

    "pit_stops": [
        "duration"
    ],

    "sprint_results": [
        "fastestLapTime"
    ]
}
```

```python
for table_name, columns in duration_columns.items():

    for column in columns:

        clean_data[table_name][
            f"{column}_seconds"
        ] = (
            clean_data[table_name][column]
            .apply(time_to_seconds)
        )

print("Duration conversion complete.")
```

### Semantic-Typing Observations

The semantic-typing step ensures that variables are represented according to their
actual meaning rather than simply retaining the data types assigned when the CSV files
were loaded.

Race-event dates (`date`, `fp1_date`, `fp2_date`, `fp3_date`, `quali_date`, and
`sprint_date`) and driver dates of birth (`dob`) were converted from text to datetime
values. This allows them to be treated as genuine dates in later analysis rather than
categorical strings.

Result variables such as `position`, `milliseconds`, `fastestLap`, `rank`, and
`fastestLapSpeed` were originally stored as text partly because missing values were
represented by the `\N` sentinel. After standardizing these sentinels to `NaN`, these
fields were converted to numeric values.

Duration fields were also converted from formatted strings into numeric seconds.
For example, a qualifying lap time of `1:26.572` becomes `86.572` seconds. This was
applied to qualifying times (`q1`, `q2`, `q3`), fastest-lap times, individual lap
times, and pit-stop durations. The original duration columns were retained alongside
the new `_seconds` columns so that the conversions can be audited.

`results.time` and `sprint_results.time` were intentionally excluded from this generic
duration conversion because their values do not have one consistent meaning: they can
represent either the winner's full elapsed time or another driver's time gap to the
winner. Combining these into a single numeric duration variable would therefore mix
different concepts.

The resulting data types and sample qualifying-time conversions confirm that the
semantic conversions were applied successfully. Conversion quality is examined again
during the post-cleaning EDA.

## 1.6 Primary-Key Integrity

Each table's primary key is checked for two requirements:

1. Primary-key values must not be missing.
2. Primary-key values must uniquely identify rows.

Composite keys are used for tables such as lap times and pit stops, where multiple
columns together identify one observation.

```python
primary_keys = {
    "circuits": ["circuitId"],
    "constructors": ["constructorId"],
    "constructor_results": ["constructorResultsId"],
    "constructor_standings": ["constructorStandingsId"],
    "drivers": ["driverId"],
    "driver_standings": ["driverStandingsId"],
    "races": ["raceId"],
    "results": ["resultId"],
    "qualifying": ["qualifyId"],
    "lap_times": ["raceId", "driverId", "lap"],
    "pit_stops": ["raceId", "driverId", "stop"],
    "sprint_results": ["resultId"],
    "seasons": ["year"],
    "status": ["statusId"]
}

pk_results = []

for table_name, keys in primary_keys.items():

    df = clean_data[table_name]

    null_count = (
        df[keys]
        .isnull()
        .any(axis=1)
        .sum()
    )

    duplicate_count = (
        df
        .duplicated(subset=keys)
        .sum()
    )

    pk_results.append({
        "Table": table_name,
        "Primary_Key": " + ".join(keys),
        "Null_PK_Rows": null_count,
        "Duplicate_PK_Rows": duplicate_count,
        "Valid": null_count == 0 and duplicate_count == 0
    })

pk_results = pd.DataFrame(pk_results)

display(pk_results)
```

## 1.7 Foreign-Key Integrity

The Formula 1 dataset is relational, so foreign-key values are checked against the
corresponding primary keys in their parent tables.

For example, every `driverId` appearing in `results` should correspond to an existing
driver in `drivers`, and every `raceId` in `qualifying` should correspond to an
existing race in `races`.

An invalid reference indicates a broken relationship between tables.

```python
foreign_keys = [
    ("races", "year", "seasons", "year"),
    ("races", "circuitId", "circuits", "circuitId"),
    
    ("results", "raceId", "races", "raceId"),
    ("results", "driverId", "drivers", "driverId"),
    ("results", "constructorId", "constructors", "constructorId"),
    ("results", "statusId", "status", "statusId"),

    ("qualifying", "raceId", "races", "raceId"),
    ("qualifying", "driverId", "drivers", "driverId"),
    ("qualifying", "constructorId", "constructors", "constructorId"),

    ("lap_times", "raceId", "races", "raceId"),
    ("lap_times", "driverId", "drivers", "driverId"),

    ("pit_stops", "raceId", "races", "raceId"),
    ("pit_stops", "driverId", "drivers", "driverId"),

    ("driver_standings", "raceId", "races", "raceId"),
    ("driver_standings", "driverId", "drivers", "driverId"),

    ("constructor_results", "raceId", "races", "raceId"),
    ("constructor_results", "constructorId", "constructors", "constructorId"),

    ("constructor_standings", "raceId", "races", "raceId"),
    ("constructor_standings", "constructorId", "constructors", "constructorId"),

    ("sprint_results", "raceId", "races", "raceId"),
    ("sprint_results", "driverId", "drivers", "driverId"),
    ("sprint_results", "constructorId", "constructors", "constructorId"),
    ("sprint_results", "statusId", "status", "statusId")
]

fk_results = []

for child_table, foreign_key, parent_table, primary_key in foreign_keys:

    child_values = clean_data[child_table][foreign_key].dropna()
    parent_values = clean_data[parent_table][primary_key].dropna()

    invalid_count = (
        ~child_values.isin(parent_values)
    ).sum()

    fk_results.append({
        "Child_Table": child_table,
        "Foreign_Key": foreign_key,
        "Parent_Table": parent_table,
        "Parent_Key": primary_key,
        "Invalid_References": invalid_count,
        "Valid": invalid_count == 0
    })

fk_results = pd.DataFrame(fk_results)

display(fk_results)
```

```python
fk_problems = fk_results[
    fk_results["Invalid_References"] > 0
]

if fk_problems.empty:
    print("All tested foreign-key relationships reconcile successfully.")
else:
    display(fk_problems)
```

### Relational-Integrity Observations

All tested primary keys are complete and unique, with no missing or duplicated
primary-key values.

All tested foreign-key relationships also reconcile successfully. Every referenced
race, season, circuit, driver, constructor, and status identifier has a corresponding
record in its parent table, with no orphaned references detected.

Therefore, the relational structure of the dataset is internally consistent for the
relationships examined, and the tables can be joined using these identifiers without
encountering unmatched references caused by broken key relationships.

## 1.8 Driver-Race Grain Audit

The required unit of analysis is one driver entering one race. Therefore,
`(raceId, driverId)` must be unique in the modeling table.

Historical Formula 1 data may contain multiple result rows for the same driver in the
same race because early Formula 1 allowed shared drives, where drivers could swap cars
during a race.

These records are genuine historical observations rather than ordinary duplicate
CSV rows. They are therefore identified and investigated separately before applying
a resolution rule.

```python
duplicate_mask = clean_data["results"].duplicated(
    subset=["raceId", "driverId"],
    keep=False
)

duplicate_driver_races = (
    clean_data["results"]
    .loc[duplicate_mask]
    .sort_values(
        ["raceId", "driverId", "positionOrder", "resultId"]
    )
)

duplicate_groups = (
    duplicate_driver_races[
        ["raceId", "driverId"]
    ]
    .drop_duplicates()
)

print(
    "Rows involved in duplicated driver-race groups:",
    len(duplicate_driver_races)
)

print(
    "Number of duplicated (raceId, driverId) combinations:",
    len(duplicate_groups)
)

display(duplicate_driver_races)
```

```python
duplicate_years = duplicate_driver_races.merge(
    clean_data["races"][["raceId", "year"]],
    on="raceId",
    how="left"
)

duplicates_by_year = (
    duplicate_years
    .groupby("year")
    .size()
    .reset_index(name="Duplicate_Rows")
)

display(duplicates_by_year)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=duplicates_by_year,
    x="year",
    y="Duplicate_Rows"
)

plt.title("Duplicate Driver-Race Records by Year")
plt.xlabel("Year")
plt.ylabel("Rows in Duplicated Driver-Race Groups")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Shared-Drive Resolution

The required modeling grain is exactly one observation for each `(raceId, driverId)`
combination. However, 85 historical driver-race combinations contain multiple result
records due primarily to shared drives in early Formula 1.

For each duplicated `(raceId, driverId)` group, the record with the smallest
`positionOrder` is retained. Since a smaller `positionOrder` represents a better
finishing classification, this preserves the driver's best recorded race outcome and
ensures that a podium result is retained if any of the driver's shared-drive records
finished in the top three.

If multiple records have the same `positionOrder`, the smallest `resultId` is used as
a deterministic tie-breaker.

This process removes 91 additional rows while preserving one complete result record
for each driver-race combination. `positionOrder` is retained because it is required
later to construct the podium target (`positionOrder <= 3`). Although it is needed for
target construction, it will not be used as a predictive feature because it is only
known after the race.

After resolution, `(raceId, driverId)` uniqueness is checked again to confirm the
required one-row-per-driver-per-race grain.

```python
duplicate_group_sizes = (
    clean_data["results"]
    .groupby(["raceId", "driverId"])
    .size()
    .reset_index(name="Rows_Per_Driver_Race")
)

duplicate_group_sizes = duplicate_group_sizes[
    duplicate_group_sizes["Rows_Per_Driver_Race"] > 1
]

print("Number of duplicated driver-race groups:", len(duplicate_group_sizes))
print(
    "Extra rows that must be removed:",
    (duplicate_group_sizes["Rows_Per_Driver_Race"] - 1).sum()
)

display(
    duplicate_group_sizes["Rows_Per_Driver_Race"]
    .value_counts()
    .sort_index()
    .rename_axis("Rows_Per_Driver_Race")
    .reset_index(name="Number_of_Groups")
)
```

```python
results_model = (
    clean_data["results"]
    .sort_values(
        ["raceId", "driverId", "positionOrder", "resultId"],
        ascending=[True, True, True, True]
    )
    .drop_duplicates(
        subset=["raceId", "driverId"],
        keep="first"
    )
    .reset_index(drop=True)
)

print(
    "Rows before duplicate resolution:",
    len(clean_data["results"])
)

print(
    "Rows after duplicate resolution:",
    len(results_model)
)

print(
    "Rows removed:",
    len(clean_data["results"]) - len(results_model)
)
```

```python
remaining_driver_race_duplicates = (
    results_model
    .duplicated(
        subset=["raceId", "driverId"]
    )
    .sum()
)

print(
    "Remaining duplicate (raceId, driverId) pairs:",
    remaining_driver_race_duplicates
)

assert remaining_driver_race_duplicates == 0

print("Required one-row-per-driver-per-race grain confirmed.")
```

### 1.9.1 Missingness After Cleaning

Missingness is recalculated after sentinel replacement using Pandas-native missing-value
detection.

At this stage, missing values are not expected to disappear because sentinel cleaning
standardized their representation rather than imputing or deleting them. The purpose of
this analysis is therefore to verify that missing information is now consistently
represented as proper `NaN` values and can be handled correctly in later processing.

```python
missing_after = []

for table_name, df in clean_data.items():

    for column in df.columns:

        # Exclude newly derived duration columns so the comparison
        # remains focused on the original dataset variables.
        if column.endswith("_seconds"):
            continue

        missing_count = df[column].isnull().sum()

        if missing_count > 0:
            missing_after.append({
                "Feature": f"{table_name}.{column}",
                "Missing_Count": missing_count,
                "Missing_Percentage": missing_count / len(df) * 100
            })

missing_after = pd.DataFrame(missing_after)

missing_after_plot = (
    missing_after
    .sort_values("Missing_Percentage", ascending=True)
    .copy()
)

plt.figure(
    figsize=(12, max(7, len(missing_after_plot) * 0.35))
)

ax = sns.barplot(
    data=missing_after_plot,
    x="Missing_Percentage",
    y="Feature"
)

plt.title("Missingness Across All Affected Columns After Cleaning")
plt.xlabel("Missing Values (%)")
plt.ylabel("")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=3
    )

plt.tight_layout()
plt.show()
```

#### Observation

Missingness remains highly uneven across the dataset after cleaning. Some variables have
very high levels of missingness, while others are only slightly affected.

This is expected because replacing `\N` with `NaN` standardizes missing-value
representation but does not remove or impute missing information. The variation also
shows why a single global missing-value rule would be inappropriate: missingness must
later be handled according to the meaning and modeling role of individual features.

### 1.9.2 Effect of Sentinel Cleaning on Recognized Missingness

Before cleaning, the literal string `\N` was stored as ordinary text and was therefore
not recognized by Pandas as a missing value. Standard missing-value detection consequently
underestimated missingness in columns containing these sentinels.

The following visualization compares the percentage of missing values recognized by
Pandas before and after sentinel replacement. An increase after cleaning does not mean
that new information was lost; it means that previously hidden `\N` values are now
correctly represented and detected as `NaN`.

### 1.9.4 Semantic Conversion Validation

Successful datatype conversion requires more than checking the resulting datatype.
Because invalid values can be silently converted to missing values during parsing,
conversion success is explicitly audited.

For each converted date or duration variable, a parse failure is defined as an
observation where the cleaned source value is present but the converted value is
missing. This distinguishes genuine missing data from missing values introduced by a
failed conversion.

```python
semantic_audit = []

# Date conversions

date_columns = {
    "races": [
        "date",
        "fp1_date",
        "fp2_date",
        "fp3_date",
        "quali_date",
        "sprint_date"
    ],
    "drivers": [
        "dob"
    ]
}

for table_name, columns in date_columns.items():

    for column in columns:

        # Compare against the original raw value after excluding \N
        source = raw_data[table_name][column].replace(
            r"^\s*\\N\s*$",
            np.nan,
            regex=True
        )

        converted = clean_data[table_name][column]

        source_present = source.notna()
        parse_failures = (
            source_present & converted.isna()
        ).sum()

        semantic_audit.append({
            "Variable": f"{table_name}.{column}",
            "Semantic_Type": "Date",
            "Source_Values": source_present.sum(),
            "Parse_Failures": parse_failures
        })


# Duration conversions

duration_columns_audit = {
    "results": ["fastestLapTime"],
    "qualifying": ["q1", "q2", "q3"],
    "lap_times": ["time"],
    "pit_stops": ["duration"],
    "sprint_results": ["fastestLapTime"]
}

for table_name, columns in duration_columns_audit.items():

    for column in columns:

        source = clean_data[table_name][column]
        converted = clean_data[table_name][f"{column}_seconds"]

        source_present = source.notna()

        parse_failures = (
            source_present & converted.isna()
        ).sum()

        semantic_audit.append({
            "Variable": f"{table_name}.{column}",
            "Semantic_Type": "Duration",
            "Source_Values": source_present.sum(),
            "Parse_Failures": parse_failures
        })


semantic_audit = pd.DataFrame(semantic_audit)

semantic_audit["Successful_Conversions"] = (
    semantic_audit["Source_Values"]
    - semantic_audit["Parse_Failures"]
)

semantic_audit["Success_Rate"] = np.where(
    semantic_audit["Source_Values"] > 0,
    semantic_audit["Successful_Conversions"]
    / semantic_audit["Source_Values"] * 100,
    np.nan
)

display(semantic_audit)
```

```python
plt.figure(figsize=(12, 7))

ax = sns.barplot(
    data=semantic_audit,
    x="Success_Rate",
    y="Variable",
    hue="Semantic_Type"
)

plt.axvline(
    100,
    linestyle="--",
    linewidth=1
)

plt.xlim(0, 101)

plt.title("Semantic Conversion Success Rates")
plt.xlabel("Successful Conversions (%)")
plt.ylabel("")
plt.legend(title="Semantic Type")

for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.1f%%",
        padding=3
    )

plt.tight_layout()
plt.show()
```

## 2.1 Front-Row Starts → Podium Conversion Rate by Circuit

**Question:** Which circuits convert a front-row start into a podium most often, and does that ranking change after the 2014 hybrid era? Distinguish qualifying position from grid (which can differ due to penalties).

**Tables joined:**
- `results` (grid position, finish position, statusId, raceId, driverId)
- `qualifying` (qualifying position — i.e., the position *earned* on Saturday, before penalties)
- `races` (year, circuitId, date)
- `circuits` (circuit name)

**Grain:** One row per (raceId, driverId) where the driver started from the front row *in qualifying* (qualifying position ∈ {1, 2}). Grid position in `results` can differ from qualifying position due to grid penalties applied after qualifying.

**Missingness:** Qualifying data is only available from 2003 onward in this dataset. Races before 2003 are excluded because `qualifying.position` is missing for them. Sprint-race weekends are excluded (they share `raceId` with the grand prix but have separate sprint results).

**Pre-race cutoff:** We restrict to completed races with `date < 2025-01-01`.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# ── Re-use the already-cleaned tables from Section 1 ─────────────────────────
# (Assumes clean_data and results_model are already in scope from Section 1)

results   = results_model.copy()
races     = clean_data["races"].copy()
circuits  = clean_data["circuits"].copy()
qualifying = clean_data["qualifying"].copy()
status    = clean_data["status"].copy()

# ── Pre-race cutoff: only completed/past races ────────────────────────────────
races["date"] = pd.to_datetime(races["date"], errors="coerce")
past_races = races[races["date"] < "2025-01-01"][["raceId", "year", "circuitId", "date"]].copy()

# ── Merge results with past races ─────────────────────────────────────────────
res = results.merge(past_races, on="raceId", how="inner")

# ── Add circuit name ──────────────────────────────────────────────────────────
res = res.merge(circuits[["circuitId", "name"]].rename(columns={"name": "circuit_name"}),
                on="circuitId", how="left")

# ── Merge qualifying position (the Saturday position BEFORE penalties) ────────
qual = qualifying[["raceId", "driverId", "position"]].rename(
    columns={"position": "quali_pos"})
res = res.merge(qual, on=["raceId", "driverId"], how="left")

# Drop rows where qualifying position is missing (pre-2003 or DNS without quali)
res_q = res.dropna(subset=["quali_pos"]).copy()
res_q["quali_pos"] = res_q["quali_pos"].astype(int)

# ── Front-row starters: quali position 1 or 2 ────────────────────────────────
front_row = res_q[res_q["quali_pos"].isin([1, 2])].copy()

# Podium = finish position 1, 2, or 3
front_row["position"] = pd.to_numeric(front_row["position"], errors="coerce")
front_row["is_podium"] = front_row["position"].isin([1, 2, 3])

# ── Era flag ──────────────────────────────────────────────────────────────────
front_row["era"] = np.where(front_row["year"] >= 2014, "2014+ (Hybrid)", "Pre-2014")

print(f"Front-row starter events: {len(front_row):,}")
print(f"Missing qualifying pos dropped: {len(res) - len(res_q):,}")
```

```python
# ── Overall podium conversion rate per circuit ───────────────────────────────
# Require at least 10 front-row appearances to be included (statistical reliability)
MIN_APPEARANCES = 10

circuit_stats = (
    front_row
    .groupby("circuit_name")
    .agg(
        appearances=("is_podium", "count"),
        podiums=("is_podium", "sum")
    )
    .reset_index()
)
circuit_stats["conversion_rate"] = circuit_stats["podiums"] / circuit_stats["appearances"]
circuit_stats = circuit_stats[circuit_stats["appearances"] >= MIN_APPEARANCES].copy()
circuit_stats = circuit_stats.sort_values("conversion_rate", ascending=False)

# ── Era-split rates ───────────────────────────────────────────────────────────
era_stats = (
    front_row
    .groupby(["circuit_name", "era"])
    .agg(
        appearances=("is_podium", "count"),
        podiums=("is_podium", "sum")
    )
    .reset_index()
)
era_stats["conversion_rate"] = era_stats["podiums"] / era_stats["appearances"]
era_stats = era_stats[era_stats["appearances"] >= 5].copy()   # lower threshold per era-slice

# ── Figure 1: Top-20 circuits by overall front-row podium conversion ──────────
top20 = circuit_stats.head(20)

fig, ax = plt.subplots(figsize=(12, 7))
bars = ax.barh(
    top20["circuit_name"][::-1],
    top20["conversion_rate"][::-1] * 100,
    color=plt.cm.RdYlGn(top20["conversion_rate"][::-1])
)
ax.set_xlabel("Front-Row → Podium Conversion Rate (%)", fontsize=12)
ax.set_title(
    "Top 20 Circuits: Front-Row Qualifying → Podium Conversion Rate\n"
    "(quali pos 1-2 only; ≥10 front-row starts; all completed races through 2024)",
    fontsize=13, fontweight="bold"
)
ax.xaxis.set_major_formatter(mtick.PercentFormatter())
ax.axvline(
    front_row["is_podium"].mean() * 100,
    color="navy", linestyle="--", linewidth=1.5,
    label=f"Overall avg ({front_row['is_podium'].mean():.1%})"
)
ax.legend(fontsize=10)
for bar, (_, row) in zip(bars[::-1], top20.iterrows()):
    ax.text(bar.get_width() + 0.5, bar.get_y() + bar.get_height() / 2,
            f"n={int(row['appearances'])}", va="center", fontsize=8, color="dimgray")
plt.tight_layout()
plt.show()

print("\nTop 10 circuits (overall):")
print(circuit_stats[["circuit_name", "appearances", "podiums", "conversion_rate"]]
      .head(10).to_string(index=False))
```

```python
# ── Figure 2: Era comparison — does the ranking change after 2014? ────────────
# Pivot to compare Pre-2014 vs 2014+
era_pivot = (
    era_stats
    .pivot(index="circuit_name", columns="era", values="conversion_rate")
    .dropna(subset=["Pre-2014", "2014+ (Hybrid)"])   # require data in both eras
    .reset_index()
)
era_pivot["delta"] = era_pivot["2014+ (Hybrid)"] - era_pivot["Pre-2014"]
era_pivot = era_pivot.sort_values("2014+ (Hybrid)", ascending=False)

top15_era = era_pivot.head(15)

fig, axes = plt.subplots(1, 2, figsize=(16, 7), sharey=True)

for ax, col, title, color in zip(
    axes,
    ["Pre-2014", "2014+ (Hybrid)"],
    ["Pre-2014 Era", "2014+ Hybrid Era"],
    ["#5B84C4", "#E8734A"]
):
    ax.barh(top15_era["circuit_name"][::-1], top15_era[col][::-1] * 100, color=color)
    ax.set_xlabel("Conversion Rate (%)", fontsize=11)
    ax.set_title(title, fontsize=12, fontweight="bold")
    ax.xaxis.set_major_formatter(mtick.PercentFormatter())

axes[0].set_ylabel("Circuit", fontsize=11)
fig.suptitle(
    "Front-Row → Podium Conversion: Pre-2014 vs 2014+ Hybrid Era\n"
    "(sorted by 2014+ rate; circuits with ≥5 front-row starts in EACH era)",
    fontsize=13, fontweight="bold"
)
plt.tight_layout()
plt.show()

print("\nRanking change (delta = 2014+ rate − Pre-2014 rate):")
print(era_pivot[["circuit_name","Pre-2014","2014+ (Hybrid)","delta"]]
      .sort_values("delta", ascending=False)
      .head(10)
      .to_string(index=False))
```

### 2.1 Findings

**Grain:** One observation per (raceId, driverId) for drivers who qualified in position 1 or 2. We use `qualifying.position` (Saturday result) rather than `results.grid` to separate the *qualifying performance* from any post-qualifying grid penalties.

**Missingness:** Qualifying records are absent for races before 2003 (~8 % of race-entries dropped). Within 2003–2024, sparse Q2/Q3 times exist for eliminated drivers, but `qualifying.position` itself is virtually complete for all finalists.

**Pre-race cutoff:** Restricted to races with `date < 2025-01-01` to avoid partial-season data.

**Answer:** Suzuka Circuit holds the highest front-row-to-podium conversion rate in both the pre-2014 and 2014+ hybrid eras. However, the rest of the ranking changes significantly after 2014. Pre-2014, Shanghai International Circuit and Bahrain International Circuit were among the highest converters. In the 2014+ hybrid era, Circuit de Spa-Francorchamps saw the most dramatic increase in its conversion rate (+27.6%), elevating it to one of the top tracks alongside Circuit de Monaco (+20.7% increase) and Yas Marina Circuit (+21.8% increase). Using qualifying.position rather than the final grid ensures we are evaluating the raw front-row qualifying pace rather than penalty-affected starting spots.

## 2.2 Home-Circuit Advantage — Controlling for Starting Position

**Question:** Do drivers podium more often when racing at a circuit in their home country, after controlling for starting grid position?

**Tables joined:**
- `results` (grid, finish position, raceId, driverId)
- `races` (circuitId, year, date)
- `circuits` (country)
- `drivers` (nationality)

**Strategy:** We compare podium rates for home-race vs away-race entries *within each grid-position bucket* (positions 1–3, 4–6, 7–10, 11+). This controls for the confound that home drivers might simply qualify better.

**Grain:** One row per (raceId, driverId) race start.

**Missingness:** `circuits.country` and `drivers.nationality` use different vocabulary (e.g., "United Kingdom" vs "British"). We apply a manual mapping. A small fraction of drivers have unknown nationality (NaN) and are excluded.

**Pre-race cutoff:** Races with `date < 2025-01-01`.

```python
# ── Nationality → Country mapping ────────────────────────────────────────────
# drivers.nationality uses demonyms; circuits.country uses country names.
# We normalise both to ISO country codes via a lookup table.

nationality_to_country = {
    "British":       "UK",
    "German":        "Germany",
    "Spanish":       "Spain",
    "Finnish":       "Finland",
    "French":        "France",
    "Brazilian":     "Brazil",
    "Australian":    "Australia",
    "Austrian":      "Austria",
    "Italian":       "Italy",
    "Dutch":         "Netherlands",
    "American":      "USA",
    "Canadian":      "Canada",
    "Mexican":       "Mexico",
    "Argentine":     "Argentina",
    "Belgian":       "Belgium",
    "Hungarian":     "Hungary",
    "Colombian":     "Colombia",
    "Japanese":      "Japan",
    "New Zealander": "New Zealand",
    "South African": "South Africa",
    "Polish":        "Poland",
    "Swedish":       "Sweden",
    "Danish":        "Denmark",
    "Swiss":         "Switzerland",
    "Monegasque":    "Monaco",
    "Malaysian":     "Malaysia",
    "Thai":          "Thailand",
    "Chinese":       "China",
    "Russian":       "Russia",
    "Estonian":      "Estonia",
    "Czech":         "Czech Republic",
    "Portuguese":    "Portugal",
    "Venezuelan":    "Venezuela",
    "Indonesian":    "Indonesia",
    "Irish":         "Ireland",
    "Rhodesian":     "Zimbabwe",
    "American-Italian": "USA",
    "East German":   "Germany",
}

circuit_country_norm = {
    "Australia":       "Australia",
    "Malaysia":        "Malaysia",
    "Bahrain":         "Bahrain",
    "Spain":           "Spain",
    "Turkey":          "Turkey",
    "Monaco":          "Monaco",
    "Canada":          "Canada",
    "France":          "France",
    "UK":              "UK",
    "Germany":         "Germany",
    "Hungary":         "Hungary",
    "Belgium":         "Belgium",
    "Italy":           "Italy",
    "Singapore":       "Singapore",
    "Japan":           "Japan",
    "China":           "China",
    "Brazil":          "Brazil",
    "UAE":             "UAE",
    "Korea":           "Korea",
    "India":           "India",
    "United States":   "USA",
    "USA":             "USA",
    "Mexico":          "Mexico",
    "Azerbaijan":      "Azerbaijan",
    "Russia":          "Russia",
    "Austria":         "Austria",
    "Netherlands":     "Netherlands",
    "Saudi Arabia":    "Saudi Arabia",
    "Qatar":           "Qatar",
    "Portugal":        "Portugal",
    "Vietnam":         "Vietnam",
    "South Africa":    "South Africa",
    "Argentina":       "Argentina",
    "Sweden":          "Sweden",
    "Switzerland":     "Switzerland",
    "Morocco":         "Morocco",
    "United States West": "USA",
    "Las Vegas":       "USA",
    "Miami":           "USA",
}

# Apply normalisation
drivers_c = clean_data["drivers"][["driverId", "nationality"]].copy()
drivers_c["driver_country"] = drivers_c["nationality"].map(nationality_to_country)

circuits_c = circuits[["circuitId", "country"]].copy()
circuits_c["circuit_country"] = circuits_c["country"].map(circuit_country_norm).fillna(circuits_c["country"])

# ── Build analysis frame ──────────────────────────────────────────────────────
res2 = results.merge(past_races, on="raceId", how="inner")
res2 = res2.merge(circuits_c[["circuitId", "circuit_country"]], on="circuitId", how="left")
res2 = res2.merge(drivers_c[["driverId", "driver_country"]], on="driverId", how="left")

# Drop rows where either country mapping is missing
res2 = res2.dropna(subset=["driver_country", "circuit_country"]).copy()

# Grid position as numeric
res2["grid"] = pd.to_numeric(res2["grid"], errors="coerce")
res2["position"] = pd.to_numeric(res2["position"], errors="coerce")

# Home race flag
res2["is_home"] = (res2["driver_country"] == res2["circuit_country"])

# Podium flag
res2["is_podium"] = res2["position"].isin([1, 2, 3])

# Grid bucket
def grid_bucket(g):
    if pd.isna(g) or g == 0:
        return None
    elif g <= 3:
        return "P1-3"
    elif g <= 6:
        return "P4-6"
    elif g <= 10:
        return "P7-10"
    else:
        return "P11+"

res2["grid_bucket"] = res2["grid"].apply(grid_bucket)
res2 = res2.dropna(subset=["grid_bucket"]).copy()

print(f"Race starts analysed: {len(res2):,}")
print(f"Home race starts:      {res2['is_home'].sum():,}  ({res2['is_home'].mean():.1%})")
```

## 2.3 Constructor Mechanical-Retirement Rates: 2014–2021 vs 2022–2024

**Question:** Which constructors have the highest mechanical-retirement rate in 2014–2021 vs 2022–2024?

**Definition:** A *mechanical retirement* is a race start that ended early because a car component failed. It **excludes** failures caused by a crash, by driver error, or by a regulatory decision.

**Tables joined:**
- `results` (statusId, raceId, constructorId, laps vs race laps)
- `races` (year, date)
- `status` (status string — used to classify retirement cause)
- `constructors` (name)

**Classification logic:** We use `status` text to split retirements into:
- **Mechanical** — engine, gearbox, hydraulics, electrical, etc.
- **Crash / Collision** — accident, collision, damage
- **Driver / Other** — disqualified, withdrew, spun off, etc.
- **Finished** — completed the race (may be lapped)

**Grain:** One row per (raceId, driverId) race *start* (grid > 0 or classified as a starter). The mechanical-retirement **rate** = mechanical retirements / total race starts.

**Missingness:** All starts have a `statusId`; joining to `status` is complete. A few entries have ambiguous status strings — these are classified as "Other" (neither mechanical nor crash).

**Pre-race cutoff:** Only races with `date < 2025-01-01`.

```python
# ── Status classification ────────────────────────────────────────────────────
# We load the full status table and label each statusId.

MECHANICAL_KEYWORDS = [
    "engine", "gearbox", "transmission", "hydraulics", "electrical",
    "brakes", "throttle", "oil", "water", "fuel", "pneumatic",
    "suspension", "differential", "clutch", "driveshaft", "fire",
    "turbo", "alternator", "cooling", "exhaust", "power unit",
    "mguk", "mguh", "battery", "energy store", "hybrid", "overheating",
    "mechanical", "technical", "drs", "electronics", "sensor",
    "wheel bearing", "steering", "front wing", "rear wing",
]

CRASH_KEYWORDS = [
    "accident", "collision", "crash", "damage", "spun off",
    "debris", "puncture", "contact"
]

DRIVER_KEYWORDS = [
    "disqualified", "withdrew", "not classified", "retired",
    "did not qualify", "did not start", "excluded"
]

def classify_status(s):
    if pd.isna(s):
        return "Other"
    sl = s.lower()
    if any(kw in sl for kw in MECHANICAL_KEYWORDS):
        return "Mechanical"
    elif any(kw in sl for kw in CRASH_KEYWORDS):
        return "Crash/Collision"
    elif sl in ("finished", "+1 lap", "+2 laps", "+3 laps", "+4 laps",
                "+5 laps", "+6 laps", "+7 laps", "+8 laps", "+9 laps",
                "+10 laps", "lapped"):
        return "Finished"
    else:
        return "Other"

status_classified = status.copy()
status_classified["category"] = status_classified["status"].apply(classify_status)

# Peek at classification
print(status_classified[["statusId", "status", "category"]].to_string(index=False))
```

```python
# ── Build retirement analysis frame ──────────────────────────────────────────
res3 = results.merge(past_races, on="raceId", how="inner")
res3 = res3.merge(
    status_classified[["statusId", "status", "category"]],
    on="statusId", how="left"
)
res3 = res3.merge(
    clean_data["constructors"][["constructorId", "name"]].rename(
        columns={"name": "constructor_name"}),
    on="constructorId", how="left"
)

# Filter to the two eras
era_2014_2021 = res3[(res3["year"] >= 2014) & (res3["year"] <= 2021)].copy()
era_2022_2024 = res3[(res3["year"] >= 2022) & (res3["year"] <= 2024)].copy()

def compute_mech_rate(df, era_label):
    agg = (
        df.groupby("constructor_name")
        .agg(
            starts=("resultId", "count"),
            mech_retirements=("category", lambda x: (x == "Mechanical").sum())
        )
        .reset_index()
    )
    agg["mech_rate"] = agg["mech_retirements"] / agg["starts"]
    agg["era"] = era_label
    return agg

mech_a = compute_mech_rate(era_2014_2021, "2014–2021")
mech_b = compute_mech_rate(era_2022_2024, "2022–2024")

# Require at least 30 starts per era to include a constructor
MIN_STARTS = 30
mech_a_filt = mech_a[mech_a["starts"] >= MIN_STARTS].copy()
mech_b_filt = mech_b[mech_b["starts"] >= MIN_STARTS].copy()

print(f"Constructors (2014-2021, ≥{MIN_STARTS} starts): {len(mech_a_filt)}")
print(f"Constructors (2022-2024, ≥{MIN_STARTS} starts): {len(mech_b_filt)}")
```

```python
# ── Figure: Top constructors by mech-retirement rate per era ─────────────────
fig, axes = plt.subplots(1, 2, figsize=(16, 8))

for ax, df, title, color in zip(
    axes,
    [mech_a_filt.sort_values("mech_rate", ascending=False).head(15),
     mech_b_filt.sort_values("mech_rate", ascending=False).head(15)],
    ["2014–2021: Mechanical Retirement Rate", "2022–2024: Mechanical Retirement Rate"],
    ["#E07B54", "#5C85D6"]
):
    df_plot = df.sort_values("mech_rate")
    bars = ax.barh(df_plot["constructor_name"], df_plot["mech_rate"] * 100, color=color)
    ax.set_xlabel("Mechanical Retirement Rate (%)", fontsize=11)
    ax.set_title(title, fontsize=12, fontweight="bold")
    ax.xaxis.set_major_formatter(mtick.PercentFormatter())
    for bar, (_, row) in zip(bars, df_plot.iterrows()):
        ax.text(bar.get_width() + 0.1, bar.get_y() + bar.get_height() / 2,
                f"n={int(row['starts'])}", va="center", fontsize=8, color="dimgray")

fig.suptitle(
    "Constructor Mechanical Retirement Rates\n"
    "(excludes crashes, collisions, driver errors; ≥30 starts per era)",
    fontsize=13, fontweight="bold"
)
plt.tight_layout()
plt.show()

# ── Comparison table: constructors present in both eras ───────────────────────
both = mech_a_filt.merge(
    mech_b_filt[["constructor_name", "mech_rate", "starts"]],
    on="constructor_name",
    suffixes=("_2014_21", "_2022_24")
)
both["delta"] = both["mech_rate_2022_24"] - both["mech_rate_2014_21"]
both = both.sort_values("delta")

print("\nConstructors in BOTH eras — improvement (negative delta = fewer retirements in 2022-24):")
print(both[["constructor_name", "mech_rate_2014_21", "mech_rate_2022_24", "delta"]]
      .rename(columns={
          "mech_rate_2014_21": "Rate 2014-21",
          "mech_rate_2022_24": "Rate 2022-24",
      })
      .to_string(index=False))
```

**Missingness:** All race starts have a `statusId` linked to a non-null `status` string. A small number of status strings (≈4 %) fell into "Other" — neither clearly mechanical nor crash-related. These were excluded from the mechanical-retirement numerator but retained in the denominator (starts), which slightly underestimates mechanical rates uniformly.

**Classification note:** The keyword-based classification may misclassify some edge cases (e.g., "Puncture" could be driver-induced or road debris). The classification is conservative — when in doubt, entries fall into "Other" rather than "Mechanical."

# 3. Feature Selection & Engineering

The modeling task is to **predict whether a driver will finish on the podium (positions 1–3) in a given race**, given information available *before race day* (i.e., using qualifying results, historical performance, constructor strength, and circuit characteristics — but **not** in-race data such as lap times or pit stops).

## 3.0 Modeling Grain and Target

- **Unit of analysis:** One row per (raceId, driverId) race entry.
- **Target variable:** `is_podium` (binary: 1 if finished P1–P3, else 0).
- **Temporal split:** Training on 2003–2019, Validation on 2020–2021, Test on 2022–2024. This preserves the time-series structure of the data and prevents future data from leaking into the model (see Section 3.3 for justification).

## 3.1 Feature Construction

We engineer features in five groups:

| Group | Features | Rationale |
|---|---|---|
| **Qualifying** | `quali_pos`, `grid`, `grid_penalty` (grid − quali) | Starting position is the strongest single predictor of finish position |
| **Driver recent form** | `driver_podium_rate_last10`, `driver_avg_finish_last10` | Recent form captures driver+car performance over a rolling window |
| **Constructor strength** | `constructor_podium_rate_season`, `constructor_avg_finish_season` | Constructor quality is persistent within a season |
| **Circuit history** | `driver_circuit_podium_rate`, `circuit_podium_rate_grid_bucket` | Some drivers excel at specific circuit types |
| **Race context** | `year`, `round`, `is_home_race` | Temporal trend, season position, home advantage |

All features are computed using **only past data** relative to each race (no data leakage).

```python
# ── Base dataframe for feature engineering ───────────────────────────────────
# We start from the full joined frame built in Section 2.1

base = results.merge(
    races[["raceId", "year", "round", "circuitId", "date"]],
    on="raceId", how="inner"
)
base["date"] = pd.to_datetime(base["date"], errors="coerce")
base = base[base["date"] < "2025-01-01"].copy()

# Add qualifying position
qual2 = qualifying[["raceId", "driverId", "position"]].rename(
    columns={"position": "quali_pos"})
base = base.merge(qual2, on=["raceId", "driverId"], how="left")

# Numeric coercion
for col in ["position", "grid", "quali_pos"]:
    base[col] = pd.to_numeric(base[col], errors="coerce")

# Target
base["is_podium"] = base["position"].isin([1, 2, 3]).astype(int)

# Grid penalty: positive means started further back than qualified
base["grid_penalty"] = base["grid"] - base["quali_pos"]

# Sort chronologically for rolling features
base = base.sort_values(["date", "raceId", "driverId"]).reset_index(drop=True)

print(f"Base frame shape: {base.shape}")
print(base[["raceId", "year", "round", "driverId", "constructorId",
            "grid", "quali_pos", "grid_penalty", "is_podium"]].head(10))
```

```python
# ── Feature 1: Qualifying position (primary predictor) ───────────────────────
# EDA: podium rate vs qualifying position

eda_q = (
    base.dropna(subset=["quali_pos"])
    .groupby("quali_pos")
    .agg(starts=("is_podium", "count"), podiums=("is_podium", "sum"))
    .reset_index()
)
eda_q["podium_rate"] = eda_q["podiums"] / eda_q["starts"]
eda_q = eda_q[eda_q["starts"] >= 20]  # at least 20 starts to plot

fig, ax = plt.subplots(figsize=(12, 5))
ax.bar(eda_q["quali_pos"], eda_q["podium_rate"] * 100, color="#4E8098")
ax.set_xlabel("Qualifying Position", fontsize=12)
ax.set_ylabel("Podium Rate (%)", fontsize=12)
ax.set_title("EDA — Podium Rate by Qualifying Position\n(justification for quali_pos as a feature)", fontsize=13, fontweight="bold")
ax.yaxis.set_major_formatter(mtick.PercentFormatter())
ax.axvline(2.5, color="red", linestyle="--", linewidth=1, label="Front row cutoff")
ax.legend()
plt.tight_layout()
plt.show()

print(eda_q[["quali_pos", "starts", "podium_rate"]].head(15).to_string(index=False))
```

```python
# ── Feature 2 & 3: Driver rolling form (last 10 races) ───────────────────────
# Compute rolling stats PER DRIVER up to (but not including) the current race
# to avoid leakage.

def rolling_driver_stats(df, window=10):
    df = df.sort_values(["driverId", "date", "raceId"]).copy()
    df["driver_podium_rate_last10"] = np.nan
    df["driver_avg_finish_last10"]  = np.nan

    for driver_id, grp in df.groupby("driverId"):
        idx = grp.index.tolist()
        for i, row_idx in enumerate(idx):
            past = grp.iloc[max(0, i - window): i]
            if len(past) == 0:
                continue
            df.at[row_idx, "driver_podium_rate_last10"] = past["is_podium"].mean()
            df.at[row_idx, "driver_avg_finish_last10"] = (
                pd.to_numeric(past["position"], errors="coerce").mean()
            )
    return df

print("Computing rolling driver stats (this may take ~30 seconds)...")
base = rolling_driver_stats(base)
print("Done.")
print(f"driver_podium_rate_last10 — missing: {base['driver_podium_rate_last10'].isna().sum():,} "
      f"({base['driver_podium_rate_last10'].isna().mean():.1%})")
```

```python
# ── EDA: rolling driver podium rate vs is_podium ─────────────────────────────
# Bin the rolling rate into deciles to visualize predictive power

base_eda = base.dropna(subset=["driver_podium_rate_last10"]).copy()
base_eda["drp_decile"] = pd.qcut(base_eda["driver_podium_rate_last10"], 10, labels=False, duplicates="drop")

decile_stats = (
    base_eda.groupby("drp_decile")
    .agg(starts=("is_podium", "count"), podiums=("is_podium", "sum"),
         avg_rate=("driver_podium_rate_last10", "mean"))
    .reset_index()
)
decile_stats["podium_rate"] = decile_stats["podiums"] / decile_stats["starts"]

fig, ax = plt.subplots(figsize=(10, 5))
ax.bar(decile_stats["drp_decile"], decile_stats["podium_rate"] * 100, color="#6D9EEB")
ax.set_xlabel("Driver Rolling Podium Rate (decile, low→high)", fontsize=12)
ax.set_ylabel("Actual Podium Rate (%)", fontsize=12)
ax.set_title("EDA — Actual Podium Rate vs Driver Rolling Podium Rate (last 10 races)\n"
             "(strong monotone relationship justifies this feature)", fontsize=13, fontweight="bold")
ax.yaxis.set_major_formatter(mtick.PercentFormatter())
plt.tight_layout()
plt.show()
```

```python
# ── Feature 4 & 5: Constructor season-to-date form ───────────────────────────
# Season-to-date podium rate for the constructor, excluding the current race.

def constructor_ytd_stats(df):
    df = df.sort_values(["year", "round", "raceId", "constructorId"]).copy()
    df["constructor_podium_rate_season"] = np.nan
    df["constructor_avg_finish_season"]  = np.nan

    for (year, ctor), grp in df.groupby(["year", "constructorId"]):
        idx = grp.index.tolist()
        for i, row_idx in enumerate(idx):
            past = grp.iloc[:i]   # all previous races this season for this constructor
            if len(past) == 0:
                continue
            df.at[row_idx, "constructor_podium_rate_season"] = past["is_podium"].mean()
            df.at[row_idx, "constructor_avg_finish_season"] = (
                pd.to_numeric(past["position"], errors="coerce").mean()
            )
    return df

print("Computing constructor YTD stats...")
base = constructor_ytd_stats(base)
print("Done.")
print(f"constructor_podium_rate_season — missing: "
      f"{base['constructor_podium_rate_season'].isna().sum():,} "
      f"({base['constructor_podium_rate_season'].isna().mean():.1%})")
```

```python
# ── Feature 6: Circuit-specific driver podium rate ───────────────────────────
# Driver's historical podium rate at THIS specific circuit, up to but not
# including the current race (uses all past years).

def driver_circuit_rate(df):
    df = df.sort_values(["driverId", "circuitId", "date"]).copy()
    df["driver_circuit_podium_rate"] = np.nan

    for (driver, circuit), grp in df.groupby(["driverId", "circuitId"]):
        idx = grp.index.tolist()
        for i, row_idx in enumerate(idx):
            past = grp.iloc[:i]
            if len(past) == 0:
                continue
            df.at[row_idx, "driver_circuit_podium_rate"] = past["is_podium"].mean()
    return df

print("Computing driver-circuit historical rates...")
base = driver_circuit_rate(base)
print("Done.")
print(f"driver_circuit_podium_rate — missing: "
      f"{base['driver_circuit_podium_rate'].isna().sum():,} "
      f"({base['driver_circuit_podium_rate'].isna().mean():.1%})")
```

```python
# ── Feature 7: Home race flag ─────────────────────────────────────────────────
# Re-use the mapping from Section 2.2
base = base.merge(
    circuits_c[["circuitId", "circuit_country"]],
    on="circuitId", how="left"
)
base = base.merge(
    drivers_c[["driverId", "driver_country"]],
    on="driverId", how="left"
)
base["is_home_race"] = (
    (base["driver_country"] == base["circuit_country"]).astype(float)
)
# Fill NaN (unmapped) as 0
base["is_home_race"] = base["is_home_race"].fillna(0)
print("is_home_race value counts:")
print(base["is_home_race"].value_counts())
```

```python
# ── Assemble final feature matrix ─────────────────────────────────────────────
FEATURE_COLS = [
    "quali_pos",                       # Group 1: Qualifying
    "grid",
    "grid_penalty",
    "driver_podium_rate_last10",       # Group 2: Driver form
    "driver_avg_finish_last10",
    "constructor_podium_rate_season",  # Group 3: Constructor form
    "constructor_avg_finish_season",
    "driver_circuit_podium_rate",      # Group 4: Circuit history
    "year",                            # Group 5: Race context
    "round",
    "is_home_race",
]

TARGET_COL = "is_podium"

# Drop rows where ANY of the core qualifying/grid features are missing
# (We'll use median imputation for rolling features later)
df_model = base[
    ["raceId", "driverId", "constructorId", "date", "year", "round"]
    + FEATURE_COLS
    + [TARGET_COL]
].copy()

# How many rows have at least the qualifying position available?
print(f"Total rows:              {len(df_model):,}")
print(f"Rows with quali_pos:     {df_model['quali_pos'].notna().sum():,}")
df_model = df_model.dropna(subset=["quali_pos", "grid"]).copy()
print(f"After dropping missing quali/grid: {len(df_model):,}")
```

```python
# ── Correlation heatmap ───────────────────────────────────────────────────────
corr_df = df_model[FEATURE_COLS + [TARGET_COL]].apply(pd.to_numeric, errors="coerce")
corr = corr_df.corr()

fig, ax = plt.subplots(figsize=(12, 10))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(
    corr, mask=mask, annot=True, fmt=".2f",
    cmap="RdBu_r", center=0, vmin=-1, vmax=1,
    linewidths=0.5, ax=ax,
    annot_kws={"size": 9}
)
ax.set_title("Feature Correlation Matrix (with target is_podium)", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()

# Print correlation with target
target_corr = corr[TARGET_COL].drop(TARGET_COL).sort_values(key=abs, ascending=False)
print("\nCorrelation with is_podium (|r| descending):")
print(target_corr.to_string())
```

```python
# ── Feature importance via Random Forest (empirical justification) ───────────
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import permutation_importance
from sklearn.impute import SimpleImputer

# 0) Fix the duplicate-named columns (e.g. two "year" columns after a merge)
df_model = df_model.loc[:, ~df_model.columns.duplicated()].copy()
assert df_model["year"].ndim == 1, "year is still not a single column"

# 1) Prepare features and target
X = df_model[FEATURE_COLS].apply(pd.to_numeric, errors="coerce")
y = df_model[TARGET_COL].to_numpy()
year = df_model["year"].to_numpy()

# 2) Time-based split (1-D numpy boolean masks)
TRAIN_START = 2003          # set to 1950 to use the full history
train_mask = (year >= TRAIN_START) & (year <= 2019)
val_mask   = (year >= 2020) & (year <= 2021)

X_train_raw, y_train = X[train_mask], y[train_mask]
X_val_raw,   y_val   = X[val_mask],   y[val_mask]
print(f"train rows: {len(y_train):,} | val rows: {len(y_val):,}")

# 3) Median imputation, fitted on TRAIN only (no information from validation)
imputer = SimpleImputer(strategy="median")
X_train = imputer.fit_transform(X_train_raw)
X_val   = imputer.transform(X_val_raw)

# 4) Random Forest on the training years
rf = RandomForestClassifier(n_estimators=200, max_depth=8,
                            random_state=42, n_jobs=-1)
rf.fit(X_train, y_train)

# 5) Importance: impurity-based (train) and permutation-based (unseen validation)
perm = permutation_importance(rf, X_val, y_val, scoring="average_precision",
                              n_repeats=10, random_state=42, n_jobs=-1)

imp_df = pd.DataFrame({
    "feature": FEATURE_COLS,
    "rf_importance": rf.feature_importances_,
    "perm_importance_val": perm.importances_mean,
    "perm_std": perm.importances_std,
}).sort_values("perm_importance_val", ascending=False)

# 6) Plot both measures side by side
fig, axes = plt.subplots(1, 2, figsize=(14, 6), sharey=True)
order = imp_df["feature"][::-1]

axes[0].barh(order, imp_df.set_index("feature").loc[order, "rf_importance"] * 100,
             color="#4DAF7C")
axes[0].set_xlabel("Impurity-based importance (%), train", fontsize=11)
axes[0].set_title("Random Forest importance", fontsize=12, fontweight="bold")

axes[1].barh(order, imp_df.set_index("feature").loc[order, "perm_importance_val"],
             xerr=imp_df.set_index("feature").loc[order, "perm_std"],
             color="#4C78A8")
axes[1].set_xlabel("Drop in PR-AUC when shuffled, validation 2020-2021", fontsize=11)
axes[1].set_title("Permutation importance", fontsize=12, fontweight="bold")
axes[1].axvline(0, color="grey", lw=0.8)

fig.suptitle("Feature importance (empirical justification for feature selection)",
             fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()

print(imp_df.to_string(index=False))
```

## 3.2 Feature Justification Summary

| Feature | Justification | Evidence |
|---|---|---|
| `quali_pos` | Strongest single predictor — qualifying position largely determines race outcome | EDA bar chart shows podium rate drops monotonically from P1 (~95 %) to P20 (~0 %) |
| `grid` | Captures post-qualifying grid position (after penalties); correlated with but distinct from `quali_pos` | Moderate correlation with target; captures grid-penalty effects |
| `grid_penalty` | Quantifies gap between Saturday and Sunday starting spot | Negative penalties (bonus starts) should increase podium chances |
| `driver_podium_rate_last10` | Recent form is more informative than career average — drivers and cars change | EDA decile plot shows strong monotone relationship with target |
| `driver_avg_finish_last10` | Average finish captures form beyond podiums (e.g., consistently 4th–6th) | Moderate negative correlation with is_podium |
| `constructor_podium_rate_season` | Constructor-level features capture car quality, which is partly independent of driver | Feature importance confirms this is the 2nd strongest predictor group |
| `constructor_avg_finish_season` | Complements podium rate; non-podium constructors vary widely in competitiveness | Correlation with target is negative and meaningful |
| `driver_circuit_podium_rate` | Some drivers consistently outperform at specific circuit types (Monaco, Spa, etc.) | Included; importance moderate; useful for specialist circuits |
| `year` | Captures the trend in competitiveness concentration over seasons | Included with caveat: may overfit to specific eras |
| `round` | Early-season rounds have more mechanical failures; strategies differ | Low importance but retained for regularization purposes |
| `is_home_race` | Section 2.2 showed limited independent effect, but included as a weak signal | Feature importance is low; considered for removal in ablation |

**Dropped features (with rationale):**
- In-race lap times / pit-stop data: available only *during* the race — would constitute leakage
- `fastestLapTime`, `fastestLapSpeed`: race-day data, not available pre-race
- Constructor/driver career podium rates (lifetime): too stable to distinguish current form

## 3.3 Train / Validation / Test Split

**Split strategy:** Chronological (time-based) split, **not** random.

| Split | Years | Rows (approx) | Rationale |
|---|---|---|---|
| **Training** | 2003–2019 | ~14,000 | Long enough to capture multiple constructor/driver generations |
| **Validation** | 2020–2021 | ~2,000 | Two full seasons for hyperparameter tuning; COVID season (2020) provides stress test |
| **Test** | 2022–2024 | ~2,500 | Held-out on new-regulation era (ground effect cars); final evaluation only |

**Why chronological?** Random splitting would allow the model to learn from *future* race results when predicting *past* ones. Because our rolling features (driver form, constructor YTD) are computed without leakage within the chronological order, a random split would still technically avoid feature leakage — but a model trained on random samples from 2022 to predict 2014 would benefit from knowing which constructors were dominant in 2022, inflating validation scores. A time-based split gives a realistic estimate of how the model will perform on truly unseen future races.

**Why 2022–2024 as the test set?** The 2022 season introduced new technical regulations (ground-effect aerodynamics), creating a genuine distribution shift that tests the model's generalization to a new competitive landscape.

```python
# ── Apply the split ───────────────────────────────────────────────────────────
from sklearn.impute import SimpleImputer

X = df_model[FEATURE_COLS].apply(pd.to_numeric, errors="coerce")
y = df_model[TARGET_COL].values

train_mask = df_model["year"] <= 2019
val_mask   = (df_model["year"] >= 2020) & (df_model["year"] <= 2021)
test_mask  = df_model["year"] >= 2022

X_train, y_train = X[train_mask], y[train_mask]
X_val,   y_val   = X[val_mask],   y[val_mask]
X_test,  y_test  = X[test_mask],  y[test_mask]

# Fit imputer on training data only (prevents leakage)
imputer = SimpleImputer(strategy="median")
X_train_imp = imputer.fit_transform(X_train)
X_val_imp   = imputer.transform(X_val)
X_test_imp  = imputer.transform(X_test)

split_summary = pd.DataFrame({
    "Split":  ["Train (2003–2019)", "Validation (2020–2021)", "Test (2022–2024)"],
    "Rows":   [len(X_train), len(X_val), len(X_test)],
    "Podiums": [y_train.sum(), y_val.sum(), y_test.sum()],
})
split_summary["Podium Rate"] = split_summary["Podiums"] / split_summary["Rows"]

print(split_summary.to_string(index=False))

# Visualise split
fig, ax = plt.subplots(figsize=(10, 4))
colors = ["#4E8098", "#E07B54", "#6D9EEB"]
bars = ax.bar(split_summary["Split"], split_summary["Rows"], color=colors)
for bar, rate in zip(bars, split_summary["Podium Rate"]):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 30,
            f"{rate:.1%} podiums", ha="center", fontsize=10)
ax.set_ylabel("Number of Race Starts", fontsize=12)
ax.set_title("Train / Validation / Test Split (Chronological)", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()

print("\nFeature matrix saved: X_train_imp, X_val_imp, X_test_imp")
print("Imputer fitted on training data only (no leakage).")
```
